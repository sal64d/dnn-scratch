```python
import numpy as np
import lib
```

```python
%matplotlib widget
import graph as g
```

```python
def sgd(w, b, x, y, epochs, eta, graph): 
    graph_loss = []
    graph_epoch = [] 
    for epoch in range(epochs):
        # feed forward
        z = w @ x + b
        a = lib.sigmoid(z)

        # it flattens around the edges making the tangent very small
        d = (a - y) * (1 - lib.sigmoid(z))  # When z is close to the edges, SP will be smoll
                                        # When z is in the middle, thn SP will be big
        # So only when we are little bit of wrong from target, the learning kicks in. 
        
        d_w = d @ x.transpose() # x = 1
        d_b = d

        w = w + (d_w * -eta)
        b = w + (d_b * -eta)

        # eval
        loss = (np.dot((y - a), (y - a)) / 2)[0][0]
        
        graph_epoch.append(epoch+1)
        graph_loss.append(loss)
        g.pltsin(graph, graph_epoch, graph_loss)
```

# Problem
Make X:1 -> Y:0

# With Cross entropy

```python
def sgd_cross_entropy(w, b, x, y, epochs, eta, graph): 
    graph_loss = []
    graph_epoch = [] 
    for epoch in range(epochs):
        # feed forward
        z = w @ x + b
        a = lib.sigmoid(z)

        d = (a - y) # with cross entropy, the sigmoid_prime term vanishes
        
        d_w = d @ x.transpose() # x = 1
        d_b = d

        w = w + (d_w * -eta)
        b = w + (d_b * -eta)

        # eval
        loss = ((y*np.log(a) + (1-y)*np.log(1-a)) / -2)[0][0]
        
        graph_epoch.append(epoch+1)
        graph_loss.append(loss)
        g.pltsin(graph, graph_epoch, graph_loss)
```

```python
graph3 = g.createplot()
```

```python
w2 = [[1.0]]
b2 = [[1.0]]
sgd_cross_entropy(w2, b2, x, y, 500, 0.1, graph3)
```

```python
w2 = [[2.0]]
b2 = [[2.0]]
sgd_cross_entropy(w2, b2, x, y, 500, 0.1, graph4)
```
