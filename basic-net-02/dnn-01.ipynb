```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
dt = pd.read_csv("../datasets/mnist_784.csv")
```

```python
def getPixel(row):
    pixels = row.iloc[np.arange(1,785)]
    floatPixels = list(map(lambda p: p.item()/255, pixels));
    return np.reshape(floatPixels, (784,1))
```

```python
def showDigit(row):
    floatPixels = getPixel(row)
    _2dPixelMap = np.reshape(floatPixels, (28,28))

    plt.imshow(_2dPixelMap, cmap="Greys")
    plt.show()
```

```python
showDigit(dt.loc[7])
```

```python
def genWeightsAndBiases(inputs, layers):
    prevNodes = inputs;
    weightsAndBiases = [];
    for nodes in layers:
        w = np.random.rand(nodes, prevNodes)
        b = np.random.rand(nodes, 1)
        prevNodes = nodes
        weightsAndBiases.append((w,b))
    return weightsAndBiases
```

```python
net = genWeightsAndBiases(784, [16, 10])
```

```python
def sigmoid(z):
    return 1/(1+np.exp(-z))
```

```python
def feedforward(weights, bias, prevActivation, activate=sigmoid):
    wx = np.matmul(weights, prevActivation)
    z = np.add(wx,bias)
    
    activation = list(map(activate, z))
    return activation
def backprop():
    return 1
```
